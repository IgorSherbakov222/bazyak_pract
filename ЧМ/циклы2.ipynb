4,1: Циклы с Заданным числом потворений

```python
for x in range(41):
    n = x**2 + x + 41
    prime = True
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            prime = False
            break
    if prime:
        print(f"x = {x}, {n} - простое число")
    else:
        print(f"x = {x}, {n} - не простое число")
```

4.2 Суммы и произведения числовых последовательностей

```python
a = float(input())
n = int(input())

S = 1 / a

for k in range(1, n):
    S += 1 / (a ** (2 * k))

print(S)
```
